# SARAI — Benchmark des modeles LLM

**Stocktaking of Arab Regional AI Initiatives**

Ce notebook evalue et compare les performances de **trois modeles LLM** utilises dans le chatbot RAG SARAI :

| Modele | Provider | Type |
|--------|----------|------|
| **Llama 3.1 8B** | Groq (API) | Cloud rapide |
| **GPT-4.1** | OpenAI (API) | Cloud haute qualite |
| **Mistral Small** | Ollama (local) | Local / prive |

**Metriques mesurees :**
- Temps de reponse (latence)
- Vitesse de generation (tokens/seconde)
- Qualite de la reponse (score LLM-as-judge 1-5)
- Pertinence / Fidelite au contexte RAG / Exactitude factuelle
- Taux d'hallucination
- Recall@k, MRR, MAP (qualite du retrieval RAG)
- Gestion des questions hors domaine

**5 categories de questions :**
1. Simples (recherche directe)
2. Complexes (multi-contraintes)
3. Multi-hop (liens entre entites)
4. Conversationnelles (contexte)
5. Hors domaine (non-SARAI)

## 4. Providers LLM pour les 3 modeles

Classes copiees depuis l'application SARAI pour encapsuler chaque modele.


````python
class LLMProvider(ABC):
    @abstractmethod
    def generate(self, system_prompt, user_query, context, image_data=None, history=None):
        pass


class OpenAIProvider(LLMProvider):
    def __init__(self):
        if not OPENAI_API_KEY:
            raise ValueError("OPENAI_API_KEY is not set")
        from openai import OpenAI as _OpenAI
        self.client = _OpenAI(api_key=OPENAI_API_KEY)
        self.model = OPENAI_MODEL

    def generate(self, system_prompt, user_query, context, image_data=None, history=None):
        try:
            content = []
            if context and context.strip():
                content.append({"type": "text", "text": f"### Context\n{context}\n\n### User Question\n{user_query}"})
            else:
                content.append({"type": "text", "text": user_query})
            messages = [{"role": "system", "content": system_prompt}]
            for h in (history or []):
                messages.append({"role": h["role"], "content": h["content"]})
            messages.append({"role": "user", "content": content})
            resp = self.client.chat.completions.create(
                model=self.model, messages=messages, temperature=DEFAULT_TEMP,
                max_tokens=MAX_TOKENS, timeout=TIMEOUT,
            )
            return resp.choices[0].message.content.strip()
        except Exception as e:
            print(f"  OpenAI error: {e}")
            return None


class GroqProvider(LLMProvider):
    def __init__(self):
        if not GROQ_API_KEY:
            raise ValueError("GROQ_API_KEY is not set")
        from openai import OpenAI
        self.client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=GROQ_API_KEY)
        self.model = GROQ_MODEL

    def generate(self, system_prompt, user_query, context, image_data=None, history=None):
        try:
            content = []
            if context and context.strip():
                content.append({"type": "text", "text": f"### Context\n{context}\n\n### User Question\n{user_query}"})
            else:
                content.append({"type": "text", "text": user_query})
            messages = [{"role": "system", "content": system_prompt}]
            for h in (history or []):
                messages.append({"role": h["role"], "content": h["content"]})
            messages.append({"role": "user", "content": content})
            resp = self.client.chat.completions.create(
                model=self.model, messages=messages, temperature=DEFAULT_TEMP,
                max_tokens=MAX_TOKENS, timeout=TIMEOUT,
            )
            return resp.choices[0].message.content.strip()
        except Exception as e:
            print(f"  Groq error: {e}")
            return None


class OllamaProvider(LLMProvider):
    def __init__(self):
        self.url = f"{OLLAMA_BASE_URL}/api/generate"
        self.model = OLLAMA_MODEL

    def generate(self, system_prompt, user_query, context, image_data=None, history=None):
        try:
            history_text = ""
            for h in (history or []):
                role = "User" if h["role"] == "user" else "Assistant"
                history_text += f"{role}: {h['content']}\n"
            prompt = f"{system_prompt}\n\n### Conversation History\n{history_text}\n\n### Context\n{context}\n\n### User Question\n{user_query}\n\n### Response"
            payload = {"model": self.model, "prompt": prompt, "stream": False,
                       "options": {"temperature": DEFAULT_TEMP}}
            resp = requests.post(self.url, json=payload, timeout=TIMEOUT)
            if resp.ok:
                return resp.json().get("response", "").strip()
        except Exception as e:
            print(f"  Ollama error: {e}")
        return None


# Instances des providers
providers = {}
if OPENAI_OK:
    providers["gpt-4.1"] = OpenAIProvider()
if GROQ_OK:
    providers["groq-llama3.1-8b"] = GroqProvider()
if OLLAMA_OK:
    providers["mistral-small"] = OllamaProvider()

print(f"Providers initialises: {', '.join(providers.keys())}")
````

## 5. Helpers (detection, formatage, prompts)

Fonctions copiees depuis l'application SARAI.


````python
GREETINGS = {"hi", "hello", "hey", "salut", "bonjour", "salam", "hiii", "helloo",
              "good morning", "good afternoon", "good evening"}

COUNTRY_KEYWORDS = ["algeria", "bahrain", "comoros", "djibouti", "egypt", "iraq",
    "jordan", "kuwait", "lebanon", "libya", "mauritania", "morocco",
    "oman", "palestine", "qatar", "saudi arabia", "somalia",
    "sudan", "syria", "tunisia", "uae", "united arab emirates", "yemen"]

TYPE_KEYWORDS = {"project": {"project", "projects", "initiative", "initiatives"},
    "stakeholder": {"stakeholder", "stakeholders", "organization", "organizations",
                     "lab", "labs", "center", "centre", "startup", "ngo"},
    "resource": {"resource", "resources", "report", "dataset", "guideline", "tool"}}

SECTOR_KEYWORDS = {"health": {"health", "healthcare", "medical", "hospital", "telemedicine"},
    "education": {"education", "school", "university", "training", "learning", "edutech"},
    "agriculture": {"agriculture", "agricultural", "farming", "food", "agritech"},
    "finance": {"finance", "financial", "banking", "fintech"},
    "energy": {"energy", "renewable", "power", "oil", "gas"},
    "governance": {"governance", "smart city", "government", "public service"},
    "environment": {"environment", "climate", "water", "environmental", "green"},
    "transportation": {"transportation", "transport", "traffic", "logistics"}}


SYSTEM_PROMPT = ("You are SARAI Assistant for the Stocktaking of Arab Regional AI Initiatives platform.\n"
    "### CRITICAL RULES:\n"
    "1. Only report database info from the context. Do not add details not present.\n"
    "2. If the database context is empty, say you don't have that information.\n"
    "3. Respond in the same language as the user (English, French, or Arabic).\n"
    "4. Be concise: 2-4 sentences unless the user asks for details.\n"
    "5. When details are requested, present them in a structured format.")


def normalize_text(text):
    return re.sub(r"\s+", " ", text.lower().strip())


def is_valid_name(name):
    if not name or len(name) < 3:
        return False
    low = name.lower()
    unique = set(low)
    if any(low.count(c) > len(name) * 0.6 for c in unique):
        return False
    if len(name) >= 6 and not set("aeiouy").intersection(low):
        return False
    return True


def format_context(results):
    lines = []
    for r in results:
        lines.append(f"[{r['type'].upper()}] {r['title']} | {r.get('country','N/A')} | {r.get('sector','N/A')}")
    return "\n".join(lines) if lines else "(empty)"


def fetch_project_details(project_id, db_conn, project_title=""):
    try:
        p = db_conn.query(Project).filter(Project.id == project_id).first()
        if not p and project_title:
            p = db_conn.query(Project).filter(Project.title.ilike(f"%{project_title}%")).first()
        if not p:
            return None
        from app.models.project import ProjectStakeholderAssociation
        from app.models.stakeholder import Stakeholder
        country_name = db_conn.query(Country.country).filter(Country.id == p.country_id).scalar() or "N/A"
        sdg_goal = db_conn.query(SDG.title).filter(SDG.id == p.sdg_id).scalar() if p.sdg_id else None
        stakeholders = (
            db_conn.query(Stakeholder.name, Stakeholder.type, ProjectStakeholderAssociation.role)
            .join(ProjectStakeholderAssociation, ProjectStakeholderAssociation.stakeholder_id == Stakeholder.id)
            .filter(ProjectStakeholderAssociation.project_id == project_id)
            .all()
        )
        return {
            "title": p.title or "",
            "organization": p.organization or "",
            "country": country_name,
            "sector": p.sector or "",
            "technology": p.technology or "",
            "description": p.description or "",
            "status": p.status or "active",
            "stakeholders": [{"name": s.name, "type": s.type, "role": s.role} for s in stakeholders],
        }
    except Exception as e:
        return None


def build_template_reply(results, query):
    if not results:
        return "I couldn't find any information matching your question in the SARAI database."
    types = {}
    for r in results:
        types.setdefault(r["type"], []).append(r)
    n = sum(len(v) for v in types.values())
    parts = [f"I found {n} relevant item(s) in the SARAI database:"]
    labels = {"project": "Projects", "stakeholder": "Organizations", "resource": "Resources"}
    for t, items in types.items():
        parts.append(f"\n**{labels.get(t, t)}:**")
        items = [it for it in items if is_valid_name(it["title"])]
        for item in items[:5]:
            parts.append(f"- {item['title']} ({item.get('country','N/A')} | {item.get('sector','N/A')})")
    return "\n".join(parts)


print("Helpers importes")
````

## 7. Fonction d'evaluation d'un modele

Pour chaque question, on execute la pipeline RAG + generation LLM et on collecte toutes les metriques.


````python
def evaluate_model(model_name, provider, questions, db_conn, verbose=True):
    """Evalue un modele sur un ensemble de questions."""
    os.chdir(str(BACKEND_DIR))
    results = []
    # Provider chain: le modele principal, puis template fallback
    from app.services.rag_service import search_rag as _search

    for i, q in enumerate(questions):
        if verbose:
            print(f"  [{i+1}/{len(questions)}] {q['id']}: {q['question'][:60]}...", end=" ")
        t0 = time.time()

        # RAG retrieval
        try:
            rag_results = _search(q["question"], db_conn, None, None, None)
            rag_results = [r for r in rag_results if is_valid_name(r.get("title", ""))]
        except Exception:
            rag_results = []

        # Format context
        ctx = format_context(rag_results)

        # LLM generation
        history = []
        if q.get("history_context"):
            for line in q["history_context"].split("\n"):
                if line.startswith("User:"):
                    history.append({"role": "user", "content": line[5:].strip()})
                elif line.startswith("Assistant:"):
                    history.append({"role": "assistant", "content": line[10:].strip()})

        gen_t0 = time.time()
        reply = provider.generate(SYSTEM_PROMPT, q["question"], ctx, None, history)
        gen_time = time.time() - gen_t0

        # Fallback template
        if not reply:
            reply = build_template_reply(rag_results, q["question"])

        total_time = time.time() - t0

        # Estimated tokens (rough: 4 chars/token)
        n_chars = len(reply)
        est_tokens = n_chars / 4
        tokens_per_sec = est_tokens / gen_time if gen_time > 0 else 0

        results.append({
            "id": q["id"],
            "category": q["category"],
            "question": q["question"],
            "ground_truth": q["ground_truth"],
            "model": model_name,
            "reply": reply,
            "n_rag_results": len(rag_results),
            "rag_titles": [r.get("title", "") for r in rag_results],
            "rag_countries": [r.get("country", "") for r in rag_results],
            "total_time_s": total_time,
            "gen_time_s": gen_time,
            "est_tokens": int(est_tokens),
            "tokens_per_sec": round(tokens_per_sec, 1),
            "provider_used": "template" if not reply else model_name,
        })
        if verbose:
            print(f"{total_time:.1f}s | {len(rag_results)} sources | {tokens_per_sec:.0f} tok/s")

    os.chdir(_old_cwd)
    return results


# Evaluation sur tous les modeles disponibles
all_results = []
print("\n=== Evaluation des modeles ===\n")
for name, prov in providers.items():
    print(f"\n--- {name} ---")
    r = evaluate_model(name, prov, BENCHMARK_QUESTIONS, db)
    all_results.extend(r)
    print(f"  Termine: {len(r)} questions")

print(f"\nTotal: {len(all_results)} evaluations collectees")
````

## 8. Auto-scoring avec LLM-as-judge

Utilisation du modele Groq (rapide) pour noter chaque reponse sur 4 axes:
1. **Pertinence** (1-5) : la reponse adresse-t-elle la question ?
2. **Fidelite au contexte** (1-5) : respecte-t-elle les donnees RAG ?
3. **Exactitude factuelle** (1-5) : les faits cites sont-ils corrects ?
4. **Hallucination** (1-5) : combien d'infos NON presentes dans le contexte ?
   (5 = aucune hallucination, 1 = beaucoup d'hallucinations)


````python
JUDGE_SYSTEM_PROMPT = ("You are an expert evaluator of RAG chatbot responses. "
    "Score the response on 4 axes from 1 (worst) to 5 (best). "
    "Return ONLY a JSON object with no explanation: "
    '{"relevance": <int>, "faithfulness": <int>, "factuality": <int>, "hallucination_free": <int>}')

JUDGE_USER_TEMPLATE = (
    "Question: {question}\n"
    "RAG Context: {context}\n"
    "Model Response: {response}\n\n"
    "Score each axis 1-5:\n"
    "1. relevance: Does the answer address the question?\n"
    "2. faithfulness: Does the answer stay true to the RAG context?\n"
    "3. factuality: Are the factual claims in the answer correct?\n"
    "4. hallucination_free: Does the answer avoid info NOT in the context? (5=none, 1=many)"
)


def score_response(question, context, response, judge_provider):
    """Score a single response using LLM-as-judge."""
    if not response or not response.strip():
        return {"relevance": 1, "faithfulness": 1, "factuality": 1, "hallucination_free": 1}
    prompt = JUDGE_USER_TEMPLATE.format(
        question=question,
        context=context[:2000] if context else "(empty)",
        response=response[:2000]
    )
    try:
        result = judge_provider.generate(JUDGE_SYSTEM_PROMPT, prompt, "")
        if result:
            # Extract JSON from response
            result = result.strip()
            # Remove markdown code fences if present
            if "```json" in result:
                result = result.split("```json")[1].split("```")[0].strip()
            elif "```" in result:
                result = result.split("```")[1].split("```")[0].strip()
            import json as _json
            scores = _json.loads(result)
            return {
                "relevance": int(scores.get("relevance", 3)),
                "faithfulness": int(scores.get("faithfulness", 3)),
                "factuality": int(scores.get("factuality", 3)),
                "hallucination_free": int(scores.get("hallucination_free", 3)),
            }
    except Exception as e:
        pass
    return {"relevance": 3, "faithfulness": 3, "factuality": 3, "hallucination_free": 3}


# Initialiser le juge (Groq, le plus rapide)
judge = None
if GROQ_OK:
    judge = GroqProvider()
    print("Juge: Groq (llama-3.1-8b-instant)")
elif OPENAI_OK:
    judge = OpenAIProvider()
    print("Juge: OpenAI (gpt-4.1)")
else:
    print("ATTENTION: aucun juge disponible, scores par defaut (3)")

# Noter toutes les reponses
if judge:
    print(f"\nNotation de {len(all_results)} reponses...")
    for i, r in enumerate(all_results):
        ctx = f"RAG returned {r['n_rag_results']} results: {', '.join(r['rag_titles'][:5])}"
        scores = score_response(r["question"], ctx, r["reply"], judge)
        r.update(scores)
        if (i + 1) % 10 == 0:
            print(f"  {i+1}/{len(all_results)} notees")
else:
    for r in all_results:
        r.update({"relevance": 3, "faithfulness": 3, "factuality": 3, "hallucination_free": 3})

print("Notation terminee")
````

## 9. Metriques de retrieval (Recall@k, MRR, MAP)

Evaluation de la qualite du RAG retrieval pour chaque question.


````python
def compute_retrieval_metrics(results):
    """Calcule Recall@k, MRR, MAP pour chaque evaluation."""
    for r in results:
        gt = r.get("ground_truth", {})
        titles = r.get("rag_titles", [])
        countries = [c.lower() for c in r.get("rag_countries", [])]

        # Determine if each retrieved item is relevant
        relevant = [False] * len(titles)
        gt_types = set(gt.get("types", []))
        gt_countries = set(c.lower() for c in gt.get("countries", []))
        gt_project = gt.get("project", "").lower()
        gt_sdg = gt.get("sdg", False)

        for i, (title, country) in enumerate(zip(titles, countries)):
            # Check by project name
            if gt_project and gt_project in title.lower():
                relevant[i] = True
            # Check by type + country
            if gt_countries and country in gt_countries:
                relevant[i] = True
            # Any result for out-of-domain should be irrelevant (ideally 0 results)
            if gt.get("type") == "out-of-domain":
                relevant[i] = False

        k_values = [1, 3, 5]
        n_rel = sum(relevant)
        r["n_relevant_total"] = n_rel
        r["relevant_mask"] = relevant

        # Recall@k
        for k in k_values:
            n_rel_at_k = sum(relevant[:k])
            r[f"recall_at_{k}"] = n_rel_at_k / max(n_rel, 1)

        # MRR
        mrr = 0.0
        for i, rel in enumerate(relevant):
            if rel:
                mrr = 1.0 / (i + 1)
                break
        r["mrr"] = mrr

        # MAP
        ap = 0.0
        if n_rel > 0:
            hits = 0
            for i, rel in enumerate(relevant):
                if rel:
                    hits += 1
                    ap += hits / (i + 1)
            ap /= n_rel
        r["map"] = ap

    return results


all_results = compute_retrieval_metrics(all_results)

# Afficher les metriques de retrieval par modele
print("\n=== Metriques de retrieval par modele ===\n")
for model in set(r["model"] for r in all_results):
    subset = [r for r in all_results if r["model"] == model]
    avg_r1 = np.mean([r["recall_at_1"] for r in subset])
    avg_r3 = np.mean([r["recall_at_3"] for r in subset])
    avg_r5 = np.mean([r["recall_at_5"] for r in subset])
    avg_mrr = np.mean([r["mrr"] for r in subset])
    avg_map = np.mean([r["map"] for r in subset])
    print(f"{model}:")
    print(f"  Recall@1: {avg_r1:.3f} | Recall@3: {avg_r3:.3f} | Recall@5: {avg_r5:.3f}")
    print(f"  MRR: {avg_mrr:.3f} | MAP: {avg_map:.3f}")
    print()
````

## 10. Aggregation des metriques par modele

Tableau comparatif final avec toutes les metriques.


````python
# Aggregation par modele
summary = []
for model in sorted(set(r["model"] for r in all_results)):
    subset = [r for r in all_results if r["model"] == model]
    s = {
        "Modele": model,
        "Latence_moyenne_s": round(np.mean([r["total_time_s"] for r in subset]), 2),
        "Latence_min_s": round(np.min([r["total_time_s"] for r in subset]), 2),
        "Latence_max_s": round(np.max([r["total_time_s"] for r in subset]), 2),
        "Tokens_sec_moy": round(np.mean([r["tokens_per_sec"] for r in subset]), 1),
        "Relevance_moy": round(np.mean([r["relevance"] for r in subset]), 2),
        "Faithfulness_moy": round(np.mean([r["faithfulness"] for r in subset]), 2),
        "Factuality_moy": round(np.mean([r["factuality"] for r in subset]), 2),
        "Hallucination_free_moy": round(np.mean([r["hallucination_free"] for r in subset]), 2),
        "Recall@1": round(np.mean([r["recall_at_1"] for r in subset]), 3),
        "Recall@3": round(np.mean([r["recall_at_3"] for r in subset]), 3),
        "Recall@5": round(np.mean([r["recall_at_5"] for r in subset]), 3),
        "MRR": round(np.mean([r["mrr"] for r in subset]), 3),
        "MAP": round(np.mean([r["map"] for r in subset]), 3),
        "Score_composite": round(np.mean([r["relevance"] + r["faithfulness"] + r["factuality"] + r["hallucination_free"]
                                           for r in subset]), 1),
    }
    summary.append(s)

df_summary = pd.DataFrame(summary)
df_summary = df_summary.sort_values("Score_composite", ascending=False)

print("\n=== TABLEAU COMPARATIF FINAL ===\n")
display(df_summary.style.format({
    "Latence_moyenne_s": "{:.2f}",
    "Latence_min_s": "{:.2f}",
    "Latence_max_s": "{:.2f}",
    "Tokens_sec_moy": "{:.1f}",
    "Relevance_moy": "{:.2f}",
    "Faithfulness_moy": "{:.2f}",
    "Factuality_moy": "{:.2f}",
    "Hallucination_free_moy": "{:.2f}",
    "Recall@1": "{:.3f}",
    "Recall@3": "{:.3f}",
    "Recall@5": "{:.3f}",
    "MRR": "{:.3f}",
    "MAP": "{:.3f}",
    "Score_composite": "{:.1f}",
}).background_gradient(cmap="RdYlGn", subset=["Score_composite", "Relevance_moy", "Faithfulness_moy"])
)
````

## 11. Analyse par categorie de questions

Comment chaque modele se comporte-t-il selon le type de question ?


````python
# Analyse par modele + categorie
cat_breakdown = []
for model in sorted(set(r["model"] for r in all_results)):
    for cat in sorted(set(r["category"] for r in all_results)):
        subset = [r for r in all_results if r["model"] == model and r["category"] == cat]
        if not subset:
            continue
        cat_breakdown.append({
            "Modele": model,
            "Categorie": cat,
            "Nb_questions": len(subset),
            "Latence_moy_s": round(np.mean([r["total_time_s"] for r in subset]), 2),
            "Tokens_sec": round(np.mean([r["tokens_per_sec"] for r in subset]), 1),
            "Relevance": round(np.mean([r["relevance"] for r in subset]), 2),
            "Faithfulness": round(np.mean([r["faithfulness"] for r in subset]), 2),
            "Factuality": round(np.mean([r["factuality"] for r in subset]), 2),
            "Hall_free": round(np.mean([r["hallucination_free"] for r in subset]), 2),
            "Recall@3": round(np.mean([r["recall_at_3"] for r in subset]), 3),
            "MRR": round(np.mean([r["mrr"] for r in subset]), 3),
        })

df_cat = pd.DataFrame(cat_breakdown)

categories_order = ["simple", "complex", "multi-hop", "conversational", "out-of-domain"]
df_cat["Categorie"] = pd.Categorical(df_cat["Categorie"], categories=categories_order, ordered=True)
df_cat = df_cat.sort_values(["Modele", "Categorie"])

print("\n=== Analyse par categorie ===\n")
display(df_cat.style.format({
    "Latence_moy_s": "{:.2f}", "Tokens_sec": "{:.1f}",
    "Relevance": "{:.2f}", "Faithfulness": "{:.2f}",
    "Factuality": "{:.2f}", "Hall_free": "{:.2f}",
    "Recall@3": "{:.3f}", "MRR": "{:.3f}",
}).background_gradient(cmap="RdYlGn", subset=["Relevance", "Faithfulness", "Factuality", "Hall_free"]))
````

## 12. Analyse des questions hors domaine

Evaluation de la capacite de chaque modele a reconnaitre qu'il ne peut pas repondre.


````python
ood = [r for r in all_results if r["category"] == "out-of-domain"]
if ood:
    ood_by_model = {}
    for r in ood:
        ood_by_model.setdefault(r["model"], []).append(r)

    print("=== Gestion des questions hors domaine ===\n")
    for model, subset in ood_by_model.items():
        # Check if model correctly refused to answer
        refusal_keywords = ["don't have", "not have", "can't", "cannot", "sorry",
                            "ne peux pas", "je n'ai pas", "desole", "désolé",
                            "no information", "not in the database", "SARAI"]
        refusals = 0
        for r in subset:
            reply_lower = r["reply"].lower()
            if any(kw in reply_lower for kw in refusal_keywords):
                refusals += 1
        refusal_rate = refusals / len(subset) * 100
        avg_len = np.mean([len(r["reply"]) for r in subset])
        print(f"{model}:")
        print(f"  Refus correct: {refusals}/{len(subset)} ({refusal_rate:.0f}%)")
        print(f"  Longueur moyenne: {avg_len:.0f} caracteres")
        # Show first response as example
        print(f"  Exemple (Q: '{subset[0]['question']}'):")
        print(f"    -> {subset[0]['reply'][:100]}...")
        print()
else:
    print("Aucune question hors domaine dans les resultats")
````
