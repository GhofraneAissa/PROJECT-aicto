## 4. Providers LLM (memes classes que l'application)

Les classes ci-dessous sont copiees depuis `app/routers/chat.py` : DeepSeekR1Provider, GroqProvider, OllamaProvider, TemplateProvider.

```python
# ── Memes classes que app/routers/chat.py ──

class LLMProvider(ABC):
    @abstractmethod
    def generate(self, system_prompt: str, user_query: str, context: str, image_data=None, history=None):
        pass

# ── DeepSeek R1 Provider ──

class DeepSeekR1Provider(LLMProvider):
    def __init__(self):
        if not GROQ_API_KEY:
            raise ValueError("GROQ_API_KEY is not set (requis pour DeepSeek via Groq)")
        from openai import OpenAI
        self.client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=GROQ_API_KEY)
        self.model = DEEPSEEK_MODEL

    def generate(self, system_prompt, user_query, context, image_data=None, history=None):
        try:
            content = []
            if context and context.strip():
                content.append({"type": "text", "text": f"### Context\n{context}\n\n### User Question\n{user_query}"})
            else:
                content.append({"type": "text", "text": user_query})
            messages = [{"role": "system", "content": system_prompt}]
            for h in (history or []):
                messages.append({"role": h["role"], "content": h["content"]})
            messages.append({"role": "user", "content": content})
            resp = self.client.chat.completions.create(
                model=self.model, messages=messages, temperature=0.1, max_tokens=1024, timeout=60,
            )
            return resp.choices[0].message.content.strip()
        except Exception as e:
            print(f"DeepSeek error: {e}")
            return None


class GroqProvider(LLMProvider):
    def __init__(self):
        if not GROQ_API_KEY:
            raise ValueError("GROQ_API_KEY is not set")
        from openai import OpenAI
        self.client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=GROQ_API_KEY)
        self.model = GROQ_MODEL
        self.vision_model = GROQ_VISION_MODEL

    def generate(self, system_prompt, user_query, context, image_data=None, history=None):
        try:
            use_vision = image_data is not None
            model = self.vision_model if use_vision else self.model
            content = []
            if context and context.strip():
                content.append({"type": "text", "text": f"### Context\n{context}\n\n### User Question\n{user_query}"})
            else:
                content.append({"type": "text", "text": user_query})
            if image_data:
                content.append({"type": "image_url", "image_url": {"url": f"data:image/jpeg;base64,{image_data}"}})
            messages = [{"role": "system", "content": system_prompt}]
            for h in (history or []):
                messages.append({"role": h["role"], "content": h["content"]})
            messages.append({"role": "user", "content": content})
            resp = self.client.chat.completions.create(
                model=model, messages=messages, temperature=0.1, max_tokens=1024, timeout=60,
            )
            return resp.choices[0].message.content.strip()
        except Exception as e:
            print(f"Groq error: {e}")
            return None

class OllamaProvider(LLMProvider):
    def __init__(self):
        self.url = f"{OLLAMA_BASE_URL}/api/generate"
        self.model = OLLAMA_MODEL

    def generate(self, system_prompt, user_query, context, image_data=None, history=None):
        try:
            history_text = ""
            for h in (history or []):
                role = "User" if h["role"] == "user" else "Assistant"
                history_text += f"{role}: {h['content']}\n"
            prompt = f"{system_prompt}\n\n### Conversation History\n{history_text}\n\n### Context\n{context}\n\n### User Question\n{user_query}\n\n### Response"
            payload = {"model": self.model, "prompt": prompt, "stream": False, "options": {"temperature": 0.3}}
            resp = requests.post(self.url, json=payload, timeout=60)
            if resp.ok:
                return resp.json().get("response", "").strip()
        except Exception as e:
            print(f"Ollama error: {e}")
        return None

class TemplateProvider(LLMProvider):
    def generate(self, system_prompt, user_query, context, image_data=None, history=None):
        return None

def build_provider_chain():
    providers = []
    # DeepSeek R1 = provider principal (via Groq API)
    if GROQ_API_KEY:
        try:
            providers.append(("deepseek-r1", DeepSeekR1Provider()))
        except Exception as e:
            print(f"DeepSeek R1 non disponible: {e}")
    # Groq = fallback distant
    if GROQ_API_KEY:
        try:
            providers.append(("groq", GroqProvider()))
        except Exception as e:
            print(f"Groq non disponible: {e}")
    if LLM_PROVIDER in ("ollama", "deepseek", "groq"):
        providers.append(("ollama", OllamaProvider()))
    providers.append(("template", TemplateProvider()))
    return providers

print("Providers initialises (DeepSeek R1 + fallback Groq/Ollama)")
```

## 5. Helpers et System Prompts (memes que l'application)

Les fonctions ci-dessous sont copiees depuis `app/routers/chat.py`.

```python
# ── Memes constantes et fonctions que app/routers/chat.py ──

GREETINGS = {"hi", "hello", "hey", "salut", "bonjour", "salam", "hiii", "helloo",
              "good morning", "good afternoon", "good evening"}

COUNTRY_KEYWORDS = ["algeria", "bahrain", "comoros", "djibouti", "egypt", "iraq",
    "jordan", "kuwait", "lebanon", "libya", "mauritania", "morocco",
    "oman", "palestine", "qatar", "saudi arabia", "somalia",
    "sudan", "syria", "tunisia", "uae", "united arab emirates", "yemen"]

TYPE_KEYWORDS = {"project": {"project", "projects", "initiative", "initiatives"},
    "stakeholder": {"stakeholder", "stakeholders", "organization", "organizations",
                     "lab", "labs", "center", "centre", "startup", "ngo"},
    "resource": {"resource", "resources", "report", "dataset", "guideline", "tool"}}

SECTOR_KEYWORDS = {"health": {"health", "healthcare", "medical", "hospital", "telemedicine"},
    "education": {"education", "school", "university", "training", "learning", "edutech"},
    "agriculture": {"agriculture", "agricultural", "farming", "food", "agritech"},
    "finance": {"finance", "financial", "banking", "fintech"},
    "energy": {"energy", "renewable", "power", "oil", "gas"},
    "governance": {"governance", "smart city", "government", "public service"},
    "environment": {"environment", "climate", "water", "environmental", "green"},
    "transportation": {"transportation", "transport", "traffic", "logistics"}}

SYSTEM_PROMPT = ("You are SARAI Assistant for the Stocktaking of Arab Regional AI Initiatives platform.\n"
    "### CRITICAL RULES — Never violate these:\n"
    "1. **ABSOLUTELY NEVER invent data.** Only report database info from the `Matching Items` section. Do not add countries, descriptions, or details not present there.\n"
    "2. Use the `Conversation Memory` section and the conversation history to remember user information (name, preferences, past questions).\n"
    "3. If the database context is empty, say \"I don't have that information in the database\" and suggest what to ask about.\n"
    "4. Respond in the same language as the user (English, French, or Arabic).\n"
    "6. **Never fabricate metrics, percentages, or quantitative results (e.g. \'20% reduction\', \'15% improvement\').** The database only stores descriptive text and does not contain any percentage metrics.\n"
    "5. When the user asks for details about a specific project, use the `Detailed Information` section. **CRITICAL: Report ONLY fields that have values.** If stakeholders, objectives, results, or metrics are not present in the section, do NOT mention them. Report the status exactly as shown (e.g. \'Draft\') without reinterpreting it. Never invent percentage numbers or quantitative results.\n"
    "### How to structure responses:\n"
    "- Start with a brief introduction mentioning the total count.\n"
    "- List items in a clear, readable format with **Title** (Country | Sector).\n"
    "- Be concise: 2-4 sentences unless the user asks for details.\n"
    "- Always cite the source type next to each item (Project, Stakeholder, Resource).\n"
    "- When details are requested, present them in a structured format with bullet points or sections.")

SYSTEM_PROMPT_FR = ("Vous etes l'assistant SARAI pour le recensement des initiatives IA dans la region arabe.\n"
    "### REGLES CRITIQUES :\n"
    "1. **N'inventez JAMAIS de donnees.** Basez-vous uniquement sur la section `Matching Items` pour les donnees.\n"
    "2. Utilisez la section `Conversation Memory` et l'historique de la conversation pour retenir les informations personnelles (prenom, preferences, questions passees).\n"
    "3. Si le contexte est insuffisant, dites \"Je n'ai pas trouve cette information dans la base de donnees.\"\n"
    "4. Repondez dans la meme langue que l'utilisateur.\n"
    "6. **N'inventez jamais de metriques, pourcentages ou resultats quantitatifs (ex: \'reduction de 20%\', \'amelioration de 15%\').** La base de donnees ne contient que du texte descriptif et aucun pourcentage.\n"
    "5. Quand l'utilisateur demande des details sur un projet specifique, utilisez la section `Detailed Information`. **CRITIQUE : ne fournissez que les champs qui ont des valeurs.** Si les parties prenantes, objectifs, resultats ou metriques ne sont pas presents dans la section, ne les mentionnez PAS. Rapportez le statut exactement comme indique (ex: \'Draft\') sans le reformuler. N'inventez jamais de pourcentages ou de resultats quantitatifs.\n"
    "### Structure des reponses :\n"
    "- Commencez par une breve introduction avec le nombre total d'elements trouves.\n"
    "- Listez les elements avec **Titre** (Pays | Secteur).\n"
    "- Soyez concis (2-4 phrases).\n"
    "- Quand des details sont demandes, presentez-les de maniere structuree avec des puces ou sections.")

def normalize_text(text):
    return re.sub(r"\s+", " ", text.lower().strip())

def is_greeting(query):
    q = normalize_text(query).rstrip("?!.")
    return q in GREETINGS or any(g == q for g in GREETINGS)

def detect_type_intent(query):
    words = set(normalize_text(query).split())
    for etype, keywords in TYPE_KEYWORDS.items():
        if words & keywords:
            return etype
    return None

def detect_country_intent(query):
    q = normalize_text(query)
    for c in COUNTRY_KEYWORDS:
        if c in q:
            return c.title()
    return None

def detect_sector_intent(query):
    q = normalize_text(query)
    for sector, keywords in SECTOR_KEYWORDS.items():
        if keywords & set(q.split()):
            return sector
    return None

def detect_language(query):
    french_markers = {'quels','quelles','quel','quelle','projets','sante','tunisie','maroc','donne','montre'}
    arabic_chars = set('\u0627\u0628\u062a\u062b\u062c\u062d\u062e\u062f\u0630\u0631\u0632\u0633\u0634\u0635\u0636\u0637\u0638\u0639\u063a\u0641\u0642\u0643\u0644\u0645\u0646\u0647\u0648\u064a\u0622\u0623\u0624\u0625\u0626')
    tokens = set(re.split(r"[\s,;:!?()]+", query.lower().strip()))
    if any(c in query for c in arabic_chars):
        return 'ar'
    if tokens & french_markers:
        return 'fr'
    return 'en'

def format_context(results):
    lines = []
    for r in results:
        lines.append(f"[{r['type'].upper()}] {r['title']} | {r.get('country','N/A')} | {r.get('sector','N/A')}")
    return "\n".join(lines) if lines else "(empty)"

def is_valid_name(name):
    if not name or len(name) < 3:
        return False
    low = name.lower()
    unique = set(low)
    # Repeated char: >60% same char
    if any(low.count(c) > len(name) * 0.6 for c in unique):
        return False
    # Gibberish >=6 chars: needs at least one vowel
    if len(name) >= 6 and not set('aeiouy').intersection(low):
        return False
    return True


DETAIL_KEYWORDS = {
    "details", "detail", "more info", "more information", "tell me more",
    "about", "describe", "description", "show", "display",
    "d\u00e9tails", "d\u00e9tail", "plus d'infos", "\u00e0 propos", "d\u00e9crire",
    "\u062a\u0641\u0627\u0635\u064a\u0644", "\u0627\u0644\u0645\u0632\u064a\u062f", "\u0648\u0635\u0641",
}


def detect_detail_intent(query):
    q = query.lower().strip()
    for kw in DETAIL_KEYWORDS:
        if kw in q:
            return True
    return False


def fetch_project_details(project_id, db, project_title=""):
    try:
        from app.models.project import Project, ProjectStakeholderAssociation
        from app.models.stakeholder import Stakeholder
        from app.models.country import Country
        from app.models.sdg import SDG
        p = db.query(Project).filter(Project.id == project_id).first()
        if not p and project_title:
            p = db.query(Project).filter(Project.title.ilike(f"%{project_title}%")).first()
        if not p:
            return None
        country_name = db.query(Country.country).filter(Country.id == p.country_id).scalar() or "N/A"
        sdg_goal = db.query(SDG.title).filter(SDG.id == p.sdg_id).scalar() if p.sdg_id else None
        stakeholders = (
            db.query(Stakeholder.name, Stakeholder.type, ProjectStakeholderAssociation.role)
            .join(ProjectStakeholderAssociation, ProjectStakeholderAssociation.stakeholder_id == Stakeholder.id)
            .filter(ProjectStakeholderAssociation.project_id == project_id)
            .all()
        )
        return {
            "title": p.title or "",
            "organization": p.organization or "",
            "country": country_name,
            "sector": p.sector or "",
            "technology": p.technology or "",
            "description": p.description or "",
            "status": p.status or "active",
            "start_date": str(p.start_date) if p.start_date else None,
            "end_date": str(p.end_date) if p.end_date else None,
            "year_of_implementation": p.year_of_implementation,
            "website": p.website or "",
            "sdg": sdg_goal or "",
            "stakeholders": [
                {"name": s.name, "type": s.type, "role": s.role}
                for s in stakeholders
            ],
        }
    except Exception as e:
        print(f"Failed to fetch project details: {e}")
        return None


def format_detail_context(results, db, max_details=3):
    detail_count = 0
    lines = []
    for r in results:
        if detail_count >= max_details:
            break
        if r["type"] == "project":
            details = fetch_project_details(r["id"], db, r.get("title", ""))
            if details:
                lines.append(f"--- Project: {details['title']} ---")
                lines.append(f"  Organization: {details['organization']}")
                lines.append(f"  Country: {details['country']}")
                lines.append(f"  Sector: {details['sector']}")
                lines.append(f"  Technology: {details['technology']}")
                lines.append(f"  Status: {details['status']}")
                if details.get("start_date"):
                    lines.append(f"  Start Date: {details['start_date']}")
                if details.get("end_date"):
                    lines.append(f"  End Date: {details['end_date']}")
                if details.get("year_of_implementation"):
                    lines.append(f"  Year: {details['year_of_implementation']}")
                if details.get("description"):
                    lines.append(f"  Description: {details['description'][:500]}")
                if details.get("website"):
                    lines.append(f"  Website: {details['website']}")
                if details.get("sdg"):
                    lines.append(f"  SDG: {details['sdg']}")
                if details["stakeholders"]:
                    lines.append("  Stakeholders:")
                    for sh in details["stakeholders"]:
                        role_str = f" ({sh['role']})" if sh['role'] else ""
                        lines.append(f"    - {sh['name']} [{sh['type']}]{role_str}")
                else:
                    lines.append("  Stakeholders: (none)")
                lines.append("")
                detail_count += 1
    return "\n".join(lines) if lines else None

def build_template_reply(results, query):
    lang = detect_language(query)
    if not results:
        msgs = {
            'fr': "Je n'ai trouve aucun resultat correspondant a votre question dans la base SARAI. Essayez de demander des projets, parties prenantes ou ressources specifiques.",
            'ar': "\u0644\u0645 \u0623\u062c\u062f \u0623\u064a \u0646\u062a\u0627\u0626\u062c \u062a\u0637\u0627\u0628\u0642 \u0633\u0624\u0627\u0644\u0643 \u0641\u064a \u0642\u0627\u0639\u062f\u0629 \u0628\u064a\u0627\u0646\u0627\u062a SARAI.",
            'en': "I couldn't find any information matching your question in the SARAI database."
        }
        return msgs.get(lang, msgs['en'])
    types = {}
    for r in results:
        types.setdefault(r["type"], []).append(r)
    n = sum(len(v) for v in types.values())
    if lang == 'fr':
        parts = [f"J'ai trouve {n} element(s) pertinent(s) dans la base SARAI :"]
        labels = {'project': 'Projets', 'stakeholder': 'Organisations', 'resource': 'Ressources'}
        for t, items in types.items():
            parts.append(f"\n**{labels.get(t, t)} :**")
            items = [it for it in items if is_valid_name(it['title'])]
            for item in items[:5]:
                parts.append(f"- {item['title']} ({item.get('country','N/A')} | {item.get('sector','N/A')})")
    elif lang == 'ar':
        parts = [f"\u0648\u062c\u062f\u062a {n} \u0646\u062a\u064a\u062c\u0629 relevant dans SARAI:"]
        for t, items in types.items():
            parts.append(f"\n**{t}:**")
            items = [it for it in items if is_valid_name(it['title'])]
            for item in items[:5]:
                parts.append(f"- {item['title']} ({item.get('country','N/A')} | {item.get('sector','N/A')})")
    else:
        parts = [f"I found {n} relevant item(s) in the SARAI database:"]
        for t, items in types.items():
            parts.append(f"\n**{t.capitalize()}s:**")
            items = [it for it in items if is_valid_name(it['title'])]
            for item in items[:5]:
                parts.append(f"- {item['title']} ({item.get('country','N/A')} | {item.get('sector','N/A')})")
    return "\n".join(parts)

print("Helpers et System Prompts importes (memes que l'application)")

```

## 5.5 Rolling Summary Memory

Mecanisme de memoire conversationnelle avec resume automatique :
- Tous les N echanges, genere un resume via LLM
- Remplace les anciens messages par le resume
- Conserve uniquement les derniers echanges (fenetre glissante)
- Reduit le nombre de tokens tout en preservant le contexte long-terme

```python
class RollingSummaryMemory:
    """Memoire conversationnelle avec resume automatique glissant."""

    def __init__(self, summary_interval=6, keep_last=2, summary_max_chars=500):
        self.history = []
        self.summary = ""
        self.exchange_count = 0
        self.summary_interval = summary_interval
        self.keep_last = keep_last
        self.summary_max_chars = summary_max_chars

    def add(self, question, reply):
        """Ajoute un echange et retourne True si un resume est necessaire."""
        self.history.append({"role": "user", "content": question})
        self.history.append({"role": "assistant", "content": reply})
        self.exchange_count += 1
        return self.exchange_count % self.summary_interval == 0

    def summarize(self):
        """Genere un resume via LLM et tronque la fenetre."""
        text = "\n".join(
            f"{h['role']}: {h['content']}" for h in self.history
        )
        last_user = next(
            (h["content"] for h in reversed(self.history) if h["role"] == "user"), ""
        )
        lang = detect_language(last_user)

        if lang == "fr":
            sys_sum = "Tu es un assistant qui resume des conversations en francais."
            sum_prompt = (
                "Resume cette conversation en 2-3 phrases en francais. "
                "Conserve le prenom de l'utilisateur, ses preferences, "
                "et tous les faits importants.\n\n"
                f"{text}\n\nResume:"
            )
        elif lang == "ar":
            sys_sum = "أنت مساعد تلخص المحادثات بالعربية."
            sum_prompt = (
                "لخص هذه المحادثة في 2-3 جمل بالعربية. "
                "احتفظ باسم المستخدم وتفضيلاته "
                "وجميع الحقائق المهمة.\n\n"
                f"{text}\n\nالملخص:"
            )
        else:
            sys_sum = "You are a conversation summarizer."
            sum_prompt = (
                "Summarize this conversation in 2-3 sentences in English. "
                "Retain the user's name, preferences, and all important facts.\n\n"
                f"{text}\n\nSummary:"
            )

        for _name, provider in build_provider_chain():
            if isinstance(provider, TemplateProvider):
                self.summary = text[: self.summary_max_chars]
                break
            s = provider.generate(sys_sum, sum_prompt, "", None)
            if s:
                self.summary = s
                break

        keep = self.keep_last * 2
        if len(self.history) > keep:
            self.history = self.history[-keep:]

    def get_summary(self):
        return self.summary

    def get_history(self):
        return self.history

    def get_context(self):
        return self.history, self.summary

    def clear(self):
        self.history = []
        self.summary = ""
        self.exchange_count = 0

    @property
    def token_estimate(self):
        total_chars = len(self.summary) + sum(
            len(h["content"]) for h in self.history
        )
        return total_chars // 4


print("RollingSummaryMemory pret (intervalle=6, fenetre=2 derniers echanges)")
```

## 6. Fonction principale (identique a l'application)

```python
RAG_HASH_FILE = None
def get_db_hash(_db):
    from sqlalchemy import func
    from app.models.project import Project
    from app.models.stakeholder import Stakeholder
    from app.models.resource import Resource
    try:
        pc = _db.query(func.count(Project.id)).scalar() or 0
        sc = _db.query(func.count(Stakeholder.id)).scalar() or 0
        rc = _db.query(func.count(Resource.id)).scalar() or 0
        mx = _db.query(func.max(Project.created_at)).scalar() or ''
        return f'{pc}|{sc}|{rc}|{mx}'
    except Exception:
        return ''

def needs_rebuild(_db):
    h = get_db_hash(_db)
    if not h:
        return False
    global RAG_HASH_FILE
    if RAG_HASH_FILE is None:
        RAG_HASH_FILE = BACKEND_DIR / '.rag_hash'
    try:
        old = RAG_HASH_FILE.read_text().strip()
        if old == h:
            return False
    except Exception:
        pass
    RAG_HASH_FILE.write_text(h)
    return True

def get_fresh_db():
    from sqlalchemy import create_engine
    from sqlalchemy.orm import sessionmaker
    _p = str(BACKEND_DIR / 'sarai.db')
    _url = f'sqlite:///{_p}'
    _eng = create_engine(_url, connect_args={'check_same_thread': False})
    return sessionmaker(bind=_eng)()

def chat(
    question: str,
    history=None,
    summary=None,
    temperature=DEFAULT_TEMP,
    top_p=0.9, top_k=40, repeat_penalty=1.1, max_tokens=1024,
) -> dict:
    """Fonction identique a chat_send() dans app/routers/chat.py"""
    t0 = time.time()

    # 1. Salutation
    if is_greeting(question):
        lang = detect_language(question)
        if lang == 'fr':
            reply = "Bonjour ! Je suis l'assistant SARAI.\n\nJe peux vous aider a trouver des projets IA, des parties prenantes et des ressources dans la region arabe."
        elif lang == 'ar':
            reply = "\u0645\u0631\u062d\u0628\u0627! \u0623\u0646\u0627 \u0645\u0633\u0627\u0639\u062f SARAI.\n\n\u064a\u0645\u0643\u0646\u0646\u064a \u0645\u0633\u0627\u0639\u062f\u062a\u0643 \u0641\u064a \u0627\u0644\u0639\u062b\u0648\u0631 \u0639\u0644\u0649 \u0645\u0634\u0627\u0631\u064a\u0639 \u0627\u0644\u0630\u0643\u0627\u0621 \u0627\u0644\u0627\u0635\u0637\u0646\u0627\u0639\u064a \u0648\u0627\u0644\u062c\u0647\u0627\u062a \u0648\u0627\u0644\u0645\u0648\u0627\u0631\u062f \u0641\u064a \u0627\u0644\u0645\u0646\u0637\u0642\u0629 \u0627\u0644\u0639\u0631\u0628\u064a\u0629."
        else:
            reply = "Hello! I am SARAI Assistant.\n\nI can help you find AI projects, stakeholders, and resources in the Arab region."
        return {"question": question, "reply": reply, "provider": "template", "time_ms": int((time.time()-t0)*1000), "n_sources": 0}

    # 2. Connexion DB fraiche + auto-rebuild RAG si donnees modifiees
    _db = get_fresh_db()
    try:
        from app.services.rag_service import search_rag as _search_rag, get_rag_service as _get_rag
        _rag_svc = _get_rag(_db)
        if needs_rebuild(_db):
            print('[AUTO] DB data changed, rebuilding RAG...')
            _rag_svc.build_index(_db)
            print(f'[AUTO] RAG rebuilt: {_rag_svc.collection.count()} documents')

        # 3. Detection d'intention
        etype = detect_type_intent(question)
        country = detect_country_intent(question)
        sector = detect_sector_intent(question)

        # 4. Recherche RAG + filtre noms invalides
        results = []
        try:
            results = _search_rag(question, _db, etype, country, sector)
            results = [r for r in results if is_valid_name(r.get('title',''))]
        except Exception:
            pass
        context = format_context(results)

        # 5. Stats DB fraiches
        try:
            from app.models.project import Project
            from app.models.stakeholder import Stakeholder
            from app.models.resource import Resource
            from sqlalchemy import func
            total_projects = _db.query(func.count(Project.id)).scalar() or 0
            total_stakeholders = _db.query(func.count(Stakeholder.id)).scalar() or 0
            total_resources = _db.query(func.count(Resource.id)).scalar() or 0
            stats = f"Total in DB: {total_projects} projects, {total_stakeholders} stakeholders, {total_resources} resources"
        except Exception:
            stats = ''

        # 6. Reponse template (toujours correcte)
        template_reply = build_template_reply(results, question)

        # 7. Detection langue + contexte detaille
        lang = detect_language(question)
        sysp = SYSTEM_PROMPT_FR if lang == 'fr' else SYSTEM_PROMPT
        detail_section = ''
        if results and detect_detail_intent(question):
            detail_text = format_detail_context(results, _db)
            if detail_text:
                detail_section = f"\n\n### Detailed Information\n{detail_text}"
        full_ctx = f"### Conversation Memory (summary of earlier exchanges)\n{summary}\n\n### Stats\n{stats}\n\n### Matching Items\n{context}{detail_section}" if summary else (f"### Stats\n{stats}\n\n### Matching Items\n{context}{detail_section}" if stats else f"### Matching Items\n{context}{detail_section}")

        # 8. Provider chain avec validation anti-hallucination
        reply = None
        active_provider = 'template'

        for name, provider in build_provider_chain():
            if isinstance(provider, TemplateProvider):
                reply = template_reply
                active_provider = name
                break
            gen = provider.generate(sysp, question, full_ctx, None, history)
            if gen:
                import re as _re
                llm_lower = gen.lower()
                db_count = len(results)
                detail_mode = detect_detail_intent(question)

                # Reject: LLM says items exist but DB has 0
                if db_count == 0 and any(w in llm_lower for w in ["there are", "i found", "voici", "j'ai trouv"]):
                    gen = None
                    continue
                # Reject: DB has items but LLM says not found
                if db_count > 0 and any(w in llm_lower for w in ["don't have that information", "no information", "not found", "aucune information"]):
                    gen = None
                    continue
                # Reject wrong counts (relaxed for detail requests)
                numbers = _re.findall(r'\d+', llm_lower)
                reject = False
                for n in numbers:
                    ni = int(n)
                    if 1 <= ni <= 100 and db_count > 0:
                        if abs(ni - db_count) >= max(2, db_count // 2) and not detail_mode:
                            reject = True
                            break
                if reject:
                    gen = None
                    continue
                # Reject percentage numbers (hallucinated metrics) even in detail_mode
                ctx_has_pct = bool(_re.search(r'\d+%', full_ctx))
                if not ctx_has_pct and _re.search(r'\d+%', gen):
                    gen = None
                    continue

                reply = gen
                active_provider = name
                break

        reply = reply or template_reply
        n_sources = len(results)
    finally:
        _db.close()

    return {"question": question, "reply": reply, "provider": active_provider, "time_ms": int((time.time()-t0)*1000), "n_sources": n_sources}

```

## 7. Test interactif

```python
print("=" * 60)
print("  SARAI - Assistant IA (Test interactif)")
print("=" * 60)
print("  Posez vos questions sur les projets IA,")
print("  les parties prenantes et les ressources")
print("  dans la region arabe.")
print("  Le chatbot retient le contexte de la conversation.")
print("  Memoire resume automatiquement tous les 6 echanges.")
print("  Fenetre glissante: 2 derniers echanges conserves.")
print("=" * 60)
print("  Tapez 'q', 'quit' ou 'exit' pour quitter.")
print("=" * 60)

memory = RollingSummaryMemory(summary_interval=6, keep_last=2)

while True:
    q = input("\nQuestion : ").strip()
    if not q:
        continue
    if q.lower() in ("q", "quit", "exit"):
        print("Au revoir !")
        break
    r = chat(q, history=memory.get_history(), summary=memory.get_summary())
    if memory.add(q, r["reply"]):
        print(f"\n--- Compression memoire apres {memory.exchange_count} echanges... ---")
        memory.summarize()
        print(f"Resume: {memory.summary[:200]}...")
        print(f"Historique tronque a {len(memory.history)} messages (derniers {memory.keep_last} echanges)")
    print(f"\nProvider: {r['provider']}")
    print(f"Reponse:\n{r['reply']}")
    print(f"\nTemps: {r['time_ms']} ms | Sources: {r['n_sources']} | Memoire: ~{memory.token_estimate} tokens")
```

## 8. Tests automatiques

```python
tests = [
    "What is SARAI?",
    "List AI projects in Tunisia",
    "AI projects in Egypt",
    "Organizations working on AI in Jordan",
    "AI resources available",
    "AI projects in healthcare",
    "Projets IA en Tunisie",
    "Organisations en Egypte",
]

for i, q in enumerate(tests, 1):
    r = chat(q)
    print(f"[{i}] {q}")
    print(f"    -> Provider: {r['provider']} | {r['n_sources']} sources | {r['time_ms']} ms")
    print(f"    -> {r['reply'][:150]}{'...' if len(r['reply'])>150 else ''}\n")
```

## 9. Performance

```python
N = 3
rows = []
for q in ["AI projects in Tunisia", "Healthcare AI", "Stakeholders in Egypt"]:
    times = [chat(q)["time_ms"] for _ in range(N)]
    rows.append({"question": q[:35], "avg_ms": np.mean(times), "min_ms": np.min(times), "max_ms": np.max(times)})

df_perf = pd.DataFrame(rows)
display(df_perf.style.format({'avg_ms': '{:.0f}', 'min_ms': '{:.0f}', 'max_ms': '{:.0f}'}))
print(f"Moyenne: {df_perf['avg_ms'].mean():.0f} ms")
```

## 10. Experimentation (temperature)

```python
exps = [("Precis", 0.1, 0.5, 20, 1.2), ("Equilibre", 0.3, 0.9, 40, 1.1),
        ("Creatif", 0.8, 0.95, 60, 1.0), ("Aleatoire", 1.2, 1.0, 100, 0.9)]
Q = "What are the main AI projects in Tunisia?"

for name, t, p, k, rp in exps:
    res = chat(Q, temperature=t, top_p=p, top_k=k, repeat_penalty=rp)
    print(f"\n[{name}] T={t} top_p={p} top_k={k} rep_pen={rp}")
    print(f"  Provider: {res['provider']} | {res['time_ms']} ms")
    print(f"  {res['reply'][:250]}")
```
